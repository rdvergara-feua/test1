```python
import threading
import time
from collections import deque

class Buffer:
    def __init__(self, capacity):
        self.queue = deque()
        self.capacity = capacity
        self.empty = threading.Semaphore(capacity)
        self.full = threading.Semaphore(0)
        self.mutex = threading.Lock()

    def produce(self, item):
        self.empty.acquire()
        with self.mutex:
            self.queue.append(item)
            print(f"Produced: {item}")
        self.full.release()

    def consume(self):
        self.full.acquire()
        with self.mutex:
            item = self.queue.popleft()
            print(f"Consumed: {item}")
        self.empty.release()
        return item

class Producer(threading.Thread):
    def __init__(self, buffer):
        super().__init__()
        self.buffer = buffer

    def run(self):
        for i in range(10):
            self.buffer.produce(i)
            time.sleep(0.1)

class Consumer(threading.Thread):
    def __init__(self, buffer):
        super().__init__()
        self.buffer = buffer

    def run(self):
        for _ in range(10):
            self.buffer.consume()
            time.sleep(0.15)

if __name__ == "__main__":
    buffer = Buffer(5)

    producer = Producer(buffer)
    consumer1 = Consumer(buffer)
    consumer2 = Consumer(buffer)

    producer.start()
    consumer1.start()
    consumer2.start()


    producer.join()
    consumer1.join()
    consumer2.join()
```
